## Part 1 — Loading & Inspecting 

````python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

#First thing always need to be done with any new dataset
print(df.shape) # (rows, columns)
print(df.head()) # First 5 rows
print(df.tail()) # Last 5 rows
print(df.columns) # columns names
print(df.dtypes) # data types per column
print(df.info()) # overview - nulls, types, memory
print(df.describe) # statistics for numeric columns
print(df.isnull().sum()) # count missing values per column
````

## Part 2 — Selecting & Filtering

````python
# Select columns
print(df['Age']) # single column -> series
print(df[['Name', 'Age', 'Survived']]) # multiple columns -> dataframe
````

````python
# Filtering rows
survived = df[df['Survived']==1]
print(survived.shape) #how many survived?
````

````python
# Multiple conditions - & means AND, | means OR, use brackets always
women_survived = df[(df['Survived']==1) & (df['Sex']== 'female')]
print(women_survived.shape)
````

````python
first_class = df[df['Pclass']==1]
print(first_class['Survived'].mean()) #survived rate in 1st class
````

````python
#.loc - label based (this is used for most things)
print(df.loc[0:4, ['Name', 'Age', 'Survived']]) #rows 0-4, specific columns
````

````python
# .iloc - position based (Used when we need row/col numbers)
print(df.iloc[0:5, 0:3]) # first 5 rows, first 3 columns
````

````python
#Value counts - great for categorical columns
print(df['Pclass'].value_counts())
print(df['Sex'].value_counts())
````

## Part 3 — GroupBy

````python
#Avg survival rate by passenger class
print(df.groupby('Pclass')['Survived'].mean())
````

````python
#Avg age by sex
print(df.groupby('Sex')['Age'].mean())
````

````python
#Multiple aggregations at once
print(df.groupby('Pclass').agg({
                 'Survived': 'mean',
                 'Age': 'mean',
                 'Fare': ['mean','max']
}))
````

````python
#Group by multiple columns
print(df.groupby(['Pclass','Sex'])['Survived'].mean())
````

````python
#Adding a new column based on existing ones
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
print(df.groupby('FamilySize')['Survived'].mean())
````

## Part 4 — Exercises

````python
# Exercise 1
# What percentage of males survived vs females?
# Use groupby — one line

print(df.groupby('Sex')['Survived'].mean())
````

````python
# Exercise 2
# Find the average fare paid by each passenger class
# Then find which class paid the most on average

print(df.groupby('Pclass')['Fare'].mean())
````

````python
# Exercise 3
# Create a new column called 'AgeGroup'
# If Age < 18 → 'Child', else → 'Adult'
# Hint: use np.where(condition, 'value_if_true', 'value_if_false')
# Then find survival rate for Children vs Adults

df['AgeGroup'] = np.where(df['Age']<18, 'Child', 'Adult')
print(df.groupby('AgeGroup')['Survived'].mean())
````

````python
# Exercise 4 (stretch)
# Find the top 5 most expensive tickets (Fare column)
# Show only their Name, Pclass, Fare, and Survived columns
# Hint: .sort_values() and .head()

df_sorted = df.sort_values('Fare')
print(df.head())
````

Here's the review:
---

Exercise 1 — Perfect ✅
Clean, correct, exactly one line. Nothing to add.

---

Exercise 2 — Correct ✅
Right approach. Just get used to reading the output — 1st class paid significantly more than 2nd and 3rd.

---

Exercise 3 — Correct logic, *one hidden problem* ⚠️
The `np.where` and groupby are right. But run this:
```python
print(df['Age'].isnull().sum())
```
Here 177 missing Age values. Those rows got labelled 'Adult' silently because `NaN < 18` is False. In real ML work this would corrupt the model. The fix:

```python
df['AgeGroup'] = np.where(df['Age'].isnull(), np.nan,
                 np.where(df['Age'] < 18, 'Child', 'Adult'))
```
This is exactly the kind of bug that causes bad models in production.

---

Exercise 4 — Bug 🐛
Two mistakes here:
```python
df_sorted = df.sort_values('Fare')  # sorts ascending — cheapest first
print(df.head())                     # printing df not df_sorted
```
Correct version:
```python
df_sorted = df.sort_values('Fare', ascending=False)
print(df_sorted[['Name', 'Pclass', 'Fare', 'Survived']].head())
```
Always double-check: are you printing the right variable? And sort direction matters — a small thing that trips up a lot of people.

---

### Overall verdict
3 out of 4 solid, one bug in Ex4 and a real-world data quality issue in Ex3 worth knowing.
