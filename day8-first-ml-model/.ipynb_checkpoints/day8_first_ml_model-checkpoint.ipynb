```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load the clean dataset from day3 or just rebuild it:
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop(columns=['Cabin'])

df['Sex_encoded'] = df['Sex'].map({'male':0, 'female':1})
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = df.apply(lambda row: 1 if (row['SibSp']+row['Parch']) ==0 else 0, axis=1)
df_clean = df[['Survived', 'Pclass', 'Sex_encoded', 'Age', 'Fare', 'IsAlone', 'FamilySize']]

print(df_clean.shape)
print(df_clean.isnull().sum())
print(df_clean.head())
```

## Part 2 — The ML Workflow

```python
#Step 1 - Seperate features (x) from target (y)

x = df_clean.drop(columns=['Survived'])  #everything except what I'm predicting
y = df_clean['Survived']

print(x.shape)  #(891, 6) - 891 passengers, 6 features
print(y.shape)  #(891,) - 891 labels (0 or 1)
```

```python
#Step 2 - Split into training and test sets

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,   #20% for testing, 80% for training
    random_state=42  #ensures same split everytime whenever code runs
)

print(x_train.shape) #(712, 6)
print(x_test.shape)  #(179, 6)
```

```python
#Step 3 - Create the model
model = LogisticRegression(max_iter=1000)
```

```python
#Step 4 - Train the model
model.fit(x_train, y_train)
```

```python
#Step 5 - Make prediction
y_pred = model.predict(x_test)
```

```python
#Step 6 - Evaluate
print(accuracy_score(y_test, y_pred))
```

## Part: 3 - Understanding the Results

```python
#Accuracy - Percentage of correct predictions
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2%}")   #Should be around 80%
```

```python
#Confusion Matrix - breaks down exactly where it went right or wrong

cm = confusion_matrix(y_test, y_pred)
fig, ax = plt.subplots(figsize=(8,6))

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Did not survive', 'Survived'],
            yticklabels=['Did not survive', 'Survived'], ax=ax)

ax.set_ylabel('Actual')
ax.set_xlabel('Predicted')
ax.set_title('Confusion Matrix')
plt.show()
```

```python
#Read it like this:
#Top left = correctly predicted "Did not Survived" (True Negative)
#Top Right = predicted "Survived" but actually didn't (False Positive)
#Botton left = predicted "Didn't Survived" but actually did (False Negative)
#Botton Right = correctly predicted "Survived" (True Positive)
```

```python
#Classification Report - precision, recall, f1 per class
print(classification_report(y_test, y_pred, target_names=['Did not survived', 'Survived']))
```

#### Understanding precision vs recall — this comes up in every ML interview:

* Precision — of all passengers you predicted survived, what % actually did? predicted positives (TP + FP)
* Recall — of all passengers who actually survived, what % did you catch? actual positives (TP + FN)
* F1 — harmonic mean of both, use this when classes are imbalanced


If you predict everyone survived, recall is 100% (you didn't miss any survivors), but precision is poor because you also labeled many non-survivors as survivors.
If you predict only one passenger survived, and that passenger truly survived, precision is 100%, but recall is very low because you missed almost all the other survivors.

## Part 2 - Feature Importance

```python
#Logistic Regression gives coefficients - these show which features
#Push the prediction toward survival (positive) or death (negative)

feature_names = x.columns
coefficients = model.coef_[0]

coef_df = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
}).sort_values('Coefficient', ascending=False)

fig, ax = plt.subplots(figsize=(10,6))
sns.barplot(data=coef_df, x='Coefficient', y='Feature',
            palette=['steelblue' if c>0 else 'salmon' for c in coef_df['Coefficient']],
            ax=ax)


ax.axvline(0, color='black', linewidth=0.8)
ax.set_title('Feature Coefficients - What drives survival predictions?')
plt.show()


#Blue bars - pushes toward survival
#Red bars - pushes away survival
```
